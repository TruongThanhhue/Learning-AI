```python
import math
```

```python
def calc_sin(x,i):
  return pow(-1,i)*(pow(x,2*i+1)/math.factorial(2*i+1))
```

```python
def calc_cos(x,i):
  return pow(-1,i)*(pow(x,2*i)/math.factorial(2*i))
```

```python
def calc_sinh(x,i):
  return (pow(x,2*i+1)/math.factorial(2*i+1))
```

```python
def calc_cosh(x,i):
  return (pow(x,2*i)/math.factorial(2*i))
```

```python
def ex4():
  n= int(input("Nhap n:"))
  x = float(input("Nhap x:"))
  sin = 0.0
  cos = 0.0
  sinh = 0.0
  cosh = 0.0
  for i in range(n+1):
    sin += calc_sin(x,i)
    cos += calc_cos(x,i)
    sinh += calc_sinh(x,i)
    cosh += calc_cosh(x,i)
  print(f'sin: {sin} \ncos {cos} \nsinh: {sinh} \ncosh: {cosh}')
ex4()
```
