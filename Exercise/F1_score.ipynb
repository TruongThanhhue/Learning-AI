```python

while True:
    try:
        tp = int(input("Nhap TP: "))
        fp = int(input("Nhap FP: "))
        fn = int(input("Nhap FN: "))

        if tp < 0 or fp < 0 or fn < 0:
            print("TP, FP, FN must be greater than or equal to 0!")
            continue

        if tp + fp == 0:
            print("Precision is undefined (TP + FP = 0)")
            continue

        if tp + fn == 0:
            print("Recall is undefined (TP + FN = 0)")
            continue

        break

    except ValueError:
        print("Please enter integers only!")

precision = tp / (tp + fp)
recall = tp / (tp + fn)

if precision + recall == 0:
    f1_score = 0
else:
    f1_score = (2 * precision * recall) / (precision + recall)

print("Precision:", precision)
print("Recall:", recall)
print("F1 Score:", f1_score)
```
