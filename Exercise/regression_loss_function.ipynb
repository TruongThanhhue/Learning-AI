```python
import random
import math
```

```python
def inside_MAE(y1,y2): ## y1 là target, y2 là predict
  return abs(y1-y2)
```

```python
def inside_MSE(y1,y2): ## y1 là target, y2 là predict
  return (y1-y2)**2
```

```python
def ex3():
  num_sample = input("Nhap n:")
  if not num_sample.isnumeric():
    print("Number must be an interger number")
    return
  num_sample = int(num_sample)
  loss_name = input('Input loss name: ')
  final = 0
  for i in range(num_sample):
    if(loss_name == 'MAE'):
      y1 = random.uniform(0,10)
      y2 = random.uniform(0,10)
      loss = inside_MAE(y1,y2)
    elif(loss_name == 'MSE' or loss_name == 'RMSE'):
      y1 = random.uniform(0,10)
      y2 = random.uniform(0,10)
      loss = inside_MSE(y1,y2)
    final += loss
    print(f'loss_name: {loss_name} sample: {i} predict: {y2} target: {y1} loss: {loss}')
  final /= num_sample
  if(loss_name == 'RMSE'):
    final = math.sqrt(final)
  print(f'Final loss: {final}')
ex3()
```
