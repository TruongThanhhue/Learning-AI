```python
import math
def is_number(n):
    try:
        float(n)   # Type-casting the string to `float`.
                   # If string is not a valid `float`,
                   # it'll raise `ValueError` exception
    except ValueError:
        return False
    return True
```

```python
def nhap():
  x= input("Nhap gia tri x:")
  is_number(x)
  if not is_number(x):
   print('x must be a number')
   return None, None
  x = float(x)
  print("sigmoid")
  print("relu")
  print("elu")
  name= input("Nhap ten activate ")
  return x, name
```

```python
import math
def sigmoid(x):
  return 1/(1+ pow(math.e,-x))
```

```python
import math
def relu(x):
  if(x<=0):
    return 0
  else:
    return x
```

```python
import math
def elu(x):
  if(x<=0):
    return 0.01*(pow(math.e,x)-1)
  else:
    return x
```

```python
import math
def ex2():
  x,name=nhap()
  if x is None:
    return
  if(name=="sigmoid"):
    print(sigmoid(x))
  elif(name=="relu"):
    print(relu(x))
  elif(name=="elu"):
    print(elu(x))
  else:
    print(name+" is not supported")
ex2()
```
