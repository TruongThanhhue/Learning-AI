$$\text{MD_nRE} = \frac{1}{m} \sum_{i=1}^{m} \left(\sqrt[n]{y_i} - \sqrt[n]{\hat{y}_i}\right)^p$$


```python
import math
```

```python
def md_nre(y, y_hat, n, p):
    y_root = y ** (1/n)
    y_hat_root = y_hat ** (1/n)
    difference = y_root - y_hat_root
    loss = difference ** p
    return loss
```

```python
md_nre(y=100, y_hat=99.5, n=2, p=1)
```
